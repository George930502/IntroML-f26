# HW3 starter: Oracle bone characters
Loads `oracle_chars.npz` (28 × 28 × 3 images; the 3 channels of a dataset image are equal) and shows a few images per character. Put `oracle_chars.npz` next to this notebook (or upload it to Colab).

```python
import numpy as np
import matplotlib.pyplot as plt

d = np.load('oracle_chars.npz')
images, labels = d['images'], d['labels']
class_names, characters = d['class_names'], d['characters']
print(images.shape, images.dtype, labels.shape)
```

```python
for c, (name, ch) in enumerate(zip(class_names, characters)):
    print(c, name, ch, (labels == c).sum())
```

```python
fig, axes = plt.subplots(10, 8, figsize=(8, 10))
for c in range(10):
    for ax, i in zip(axes[c], np.flatnonzero(labels == c)[:8]):
        ax.imshow(images[i], cmap='gray', vmin=0, vmax=255)
    for ax in axes[c]:
        ax.axis('off')
    axes[c, 0].set_title(class_names[c], fontsize=8, loc='left')
plt.tight_layout()
plt.show()
```

## Section 2 Question 5: saving your handwriting
Crop a square around each of your 30 characters (any image tool) and save the crops as image files. `save_handwriting` resizes them to 28 × 28, keeps RGB, checks the format, and writes `HW3_handwriting_{student_id}.npz`.

```python
from PIL import Image

def check_handwriting(images, labels):
    assert images.shape == (30, 28, 28, 3), f'images must be 30 x 28 x 28 x 3, got {images.shape}'
    assert images.dtype == np.uint8, f'images must be uint8, got {images.dtype}'
    assert (labels == np.repeat(np.arange(10), 3)).all(), 'labels must be 0,0,0,1,...,9'

def save_handwriting(paths, student_id):
    """paths: your 30 square crops, 3 per character, in label order 0, 0, 0, 1, ..., 9."""
    assert len(paths) == 30, f'need 30 images, got {len(paths)}'
    images = []
    for p in paths:
        im = Image.open(p).convert('RGB')  # keep the colors: no grayscale, no inversion
        assert im.width == im.height, f'{p} is {im.width} x {im.height}; crop a square first'
        images.append(np.asarray(im.resize((28, 28), Image.LANCZOS)))
    images, labels = np.stack(images).astype(np.uint8), np.repeat(np.arange(10), 3)
    check_handwriting(images, labels)
    np.savez_compressed(f'HW3_handwriting_{student_id}.npz', images=images, labels=labels)

# Example: save_handwriting([f'my_chars/{c}_{k}.png' for c in range(10) for k in range(1, 4)], 113000000)
# Check a saved file:
# h = np.load('HW3_handwriting_113000000.npz'); check_handwriting(h['images'], h['labels'])
```
